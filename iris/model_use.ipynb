# Classificator Use to Predict Iris Class

```python
import numpy as np
import pandas as pd

from keras.models import load_model
from scikeras.wrappers import KerasClassifier
from sklearn.metrics import classification_report
```

## Load Model

```python
classifier:KerasClassifier = load_model("models/classifier.keras")
classifier.summary()
```

## Use Classifier

```python
df = pd.read_csv("datasets/iris.csv")
display(df)

X = df[["sepal length", "sepal width", "petal length", "petal width"]]
y = df[["class"]]
```

### Predict Translation

```python
def translate_probability_prediction(classifier:KerasClassifier, X:pd.DataFrame) -> pd.DataFrame:
    predictions:list[dict[str, float]] = []
    for pred in classifier.predict(X):
        predictions.append({
            "Iris-setosa": pred[0],
            "Iris-versicolor": pred[1],
            "Iris-virginica": pred[2]
        })

    return pd.DataFrame(predictions)

def translate_categorical_prediction(classifier:KerasClassifier, X:pd.DataFrame) -> pd.DataFrame:
    translator:dict[int, str] = {
        0: "Iris-setosa",
        1: "Iris-versicolor",
        2: "Iris-virginica"
    }

    predictions:list[dict[str, str]] = []
    for pred in classifier.predict(X):
        predictions.append({
            "class": translator[np.argmax(pred)]
        })

    return pd.DataFrame(predictions)
```
