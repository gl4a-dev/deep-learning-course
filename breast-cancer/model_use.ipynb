# Classificator Use to Predict Breast Cancer

```python
import pandas as pd

from keras.models import load_model
from scikeras.wrappers import KerasClassifier
from sklearn.metrics import classification_report
```

## Load Model

```python
classificator:KerasClassifier = load_model("models/classificator.keras")
classificator.summary()
```

## Use Classifier

```python
X = pd.read_csv("datasets/cancer-features.csv")
y = pd.read_csv("datasets/cancer-target.csv")

display(X)
display(y)
```

### Evaluation

```python
benign_pred = classificator.predict(X) > 0.5
benign_true = y.values > 0.5

report = classification_report(
    y_true=benign_true,
    y_pred=benign_pred
)
print(report)
```
